**A. Membangun Katalog (Create)** *(bobot 20%)*

Buat database baru bernama `kopi_nusantara`, collection bernama `menu`. Sisipkan **minimal 12 dokumen** menu dengan field bervariasi sesuai kategorinya:
- Menu kategori **"Kopi"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `tingkat_kepahitan` (skala 1-5).
- Menu kategori **"Makanan"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `pedas_level` (skala 0-3).
- Boleh menambahkan field lain sesuai kreativitas anda (mis. `stok`, `rekomendasi_barista`).

```python
from pymongo import MongoClient

# 1. Koneksi ke MongoDB & inisialisasi Database/Collection
client = MongoClient("mongodb://localhost:27017/")
db_tugas = client["kopi_nusantara"]
koleksi_menu = db_tugas["menu"]

# Bersihkan collection jika sudah pernah ada
koleksi_menu.delete_many({})

# 2. Menyisipkan 12 dokumen menu dengan atribut bervariasi
data_menu = [
    # Kategori Kopi (Wajib: nama_menu, harga, kategori, tingkat_kepahitan)
    {"nama_menu": "Espresso Always Single", "kategori": "Kopi", "harga": 18000, "tingkat_kepahitan": 5, "stok": 50},
    {"nama_menu": "Americano Beku", "kategori": "Kopi", "harga": 22000, "tingkat_kepahitan": 4, "stok": 80},
    {"nama_menu": "Kopi Sugar Aren", "kategori": "Kopi", "harga": 20000, "tingkat_kepahitan": 2, "stok": 100, "rekomendasi_barista": True},
    {"nama_menu": "Piccolo Latte", "kategori": "Kopi", "harga": 24000, "tingkat_kepahitan": 3, "stok": 40},
    {"nama_menu": "Cold Brew Aceh Gayo", "kategori": "Kopi", "harga": 28000, "tingkat_kepahitan": 5, "stok": 30, "rekomendasi_barista": True},
    
    # Kategori Makanan (Wajib: nama_menu, harga, kategori, pedas_level)
    {"nama_menu": "Ayam Geprek Sambal Kicau", "kategori": "Makanan", "harga": 25000, "pedas_level": 3, "stok": 45},
    {"nama_menu": "Mie Goreng Jawa", "kategori": "Makanan", "harga": 22000, "pedas_level": 1, "stok": 60},
    {"nama_menu": "Nasi Goreng Krezy", "kategori": "Makanan", "harga": 27000, "pedas_level": 2, "stok": 35},
    {"nama_menu": "Singkong Goreng Keju", "kategori": "Makanan", "harga": 15000, "pedas_level": 0, "stok": 70},
    
    # Kategori Non-Kopi / Dessert (Kreativitas tambahan)
    {"nama_menu": "Matcha Ice Latte", "kategori": "Non-Kopi", "harga": 25000, "stok": 50},
    {"nama_menu": "Thai Tea", "kategori": "Non-Kopi", "harga": 16000, "stok": 90},
    {"nama_menu": "Choco Magma Cake", "kategori": "Dessert", "harga": 20000, "stok": 25}
]

koleksi_menu.insert_many(data_menu)
print("Jumlah menu tersimpan:", koleksi_menu.count_documents({}))
```

**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :
1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.
2. Seluruh menu dengan `harga` di antara 15000 dan 30000.
3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
# Kueri 1: Kopi dengan tingkat kepahitan > 3
print("--- 1. Menu Kopi dengan Tingkat Kepahitan > 3 ---")
for m in koleksi_menu.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}}):
    print(f"- {m['nama_menu']} | Kepahitan: {m['tingkat_kepahitan']}")

print()

# Kueri 2: Menu dengan harga di antara 15.000 dan 30.000
print("--- 2. Menu dengan Harga Rp15.000 - Rp30.000 ---")
for m in koleksi_menu.find({"harga": {"$gte": 15000, "$lte": 30000}}):
    print(f"- {m['nama_menu']} | Harga: Rp{m['harga']:,}")

print()

# Kueri 3: Menu yang TIDAK memiliki field 'pedas_level'
print("--- 3. Menu yang Tidak Memiliki Field 'pedas_level' ---")
for m in koleksi_menu.find({"pedas_level": {"$exists": False}}):
    print(f"- {m['nama_menu']} ({m['kategori']})")
```

**C. Pembaruan Data (Update)** *(bobot 15%)*

Tambahkan field `status` bernilai `"Tersedia"` ke seluruh dokumen menggunakan `update_many()`. Kemudian, ubah `status` menjadi `"Habis"` khusus untuk 2 menu pilihan anda menggunakan `update_one()`.

```python
# 1. Tambah field status="Tersedia" ke SELURUH dokumen
koleksi_menu.update_many({}, {"$set": {"status": "Tersedia"}})

# 2. Ubah status menjadi "Habis" untuk 2 menu pilihan (contoh: Cold Brew & Ayam Geprek)
koleksi_menu.update_one({"nama_menu": "Cold Brew Aceh Gayo"}, {"$set": {"status": "Habis"}})
koleksi_menu.update_one({"nama_menu": "Ayam Geprek Sambal Kicau"}, {"$set": {"status": "Habis"}})

# Tampilkan hasil update
print("--- Hasil Pembaruan Status Menu ---")
for m in koleksi_menu.find({}, {"_id": 0, "nama_menu": 1, "kategori": 1, "status": 1}):
    print(f"- {m['nama_menu']} ({m['kategori']}): {m['status']}")
```

**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (ikuti pola Sub-bab 9.6 — perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.
2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.

```python
from pyspark.sql import SparkSession
import pandas as pd
from pyspark.sql.functions import avg, count

# Inisialisasi SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

# 1. Ambil dokumen dari MongoDB
menu_docs = list(koleksi_menu.find({}))

# 2. Konversi ke Pandas DataFrame
df_pandas_menu = pd.DataFrame(menu_docs)
df_pandas_menu["_id"] = df_pandas_menu["_id"].astype(str)

# --- SOLUSI: Tangani kolom boolean/opsional yang memiliki NaN ---
# Jika kolom 'rekomendasi_barista' ada, isi NaN dengan False agar tipe datanya konsisten (Boolean)
if "rekomendasi_barista" in df_pandas_menu.columns:
    df_pandas_menu["rekomendasi_barista"] = df_pandas_menu["rekomendasi_barista"].fillna(False).astype(bool)

# Lakukan hal serupa jika ada kolom opsional lain yang bertipe boolean
# -------------------------------------------------------------

# 3. Konversi ke Spark DataFrame
df_spark_menu = spark.createDataFrame(df_pandas_menu)

# Output 1: Tampilkan Schema
print("--- Schema Spark DataFrame ---")
df_spark_menu.printSchema()

print("--- Tampilan Data Spark DataFrame ---")
df_spark_menu.show(truncate=False)

# Output 2: Rata-rata harga per kategori menggunakan groupBy
print("--- Rata-rata Harga per Kategori ---")
df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show()

spark.stop()
```

**E. Kesimpulan** *(bobot 20%)*

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?


Berdasarkan praktikum dan tugas mandiri yang telah dilakukan, tantangan utama saat mengonversi data dari MongoDB ke Spark DataFrame adalah perbedaan filosofi pengelolaan skema data. MongoDB menerapkan konsep flexible schema di mana setiap dokumen di dalam satu collection dapat memiliki atribut yang tidak seragam—seperti atribut tingkat_kepahitan pada menu kopi atau rekomendasi_barista yang hanya dimiliki oleh menu tertentu. Sebaliknya, Apache Spark menuntut struktur skema tabel yang kaku dan seragam (rigid schema) agar pemrosesan terdistribusi dapat berjalan efisien.   

Dalam implementasi praktisnya, perbedaan ini memicu tantangan teknis berupa error tipe data seperti [CANNOT_MERGE_TYPE] Can not merge type DoubleType and BooleanType. Kendala tersebut muncul karena Pandas secara otomatis mengisi nilai kosong (missing values) pada kolom opsional dengan nilai NaN (yang terbaca sebagai tipe data float/double), sehingga berbenturan dengan tipe data Boolean (True) dari dokumen lain saat Spark melakukan inferensi skema otomatis.

Cara mengatasi tantangan ini adalah dengan melakukan pembersihan data (data cleaning) secara proaktif di tingkat Pandas DataFrame sebelum memanggil fungsi spark.createDataFrame(). Kita dapat menggunakan metode penanganan nilai kosong, seperti .fillna(False).astype(bool), untuk menyeragamkan kolom boolean yang opsional. Dengan penanganan yang tepat, konversi data NoSQL ke Spark DataFrame dapat berjalan mulus tanpa error, sehingga data katalog yang dinamis tetap bisa dianalisis secara akurat menggunakan fungsi agregasi seperti groupBy di PySpark.
