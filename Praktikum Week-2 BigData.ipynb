```python
import pandas as pd

# 1. Buat data transaksi
data_penjualan = {
    "kota": ["Magelang", "Yogyakarta", "Semarang", "Solo", "Magelang", "Yogyakarta", "Semarang", "Solo"],
    "minggu": [1, 1, 1, 1, 2, 2, 2, 2],
    "produk": ["Kaos", "Kaos", "Kaos", "Kaos", "Kaos", "Kaos", "Kaos", "Kaos"],
    "unit_terjual": [120, 95, 150, 80, 135, 110, 160, 90],
    "harga_satuan": [75000, 75000, 75000, 75000, 75000, 75000, 75000, 75000]
}

df = pd.DataFrame(data_penjualan)

# 2. Wajib jalankan baris ini untuk membuat kolom 'total_pendapatan'
df["total_pendapatan"] = df["unit_terjual"] * df["harga_satuan"]

# 3. Lakukan Group By setelah kolom terbuat
total_per_kota = df.groupby("kota")["total_pendapatan"].sum().sort_values(ascending=False)
print("Total Pendapatan Per Kota:")
print(total_per_kota)

# 4. Filtering: menampilkan hanya data dari kota Magelang
data_magelang = df[df["kota"] == "Magelang"]
print("\n",data_magelang)

# 5. Filter data: Tampilkan yang laku lebih dari 100 pcs di Minggu ke-2
data_laris = df[(df["unit_terjual"] > 100) & (df["minggu"] == 2)]
print("\n",data_laris)
```

```python
# Jawaban Soal 1 di sini
import pandas as pd

# 1. Bikin data 5 mahasiswa
data_nilai = {
    "nama": ["Ahmad", "Siti", "Budi", "Dewi", "Eko"],
    "uts": [75, 88, 60, 95, 82],
    "uas": [80, 92, 70, 90, 85]
}

# 2. Bikin DataFrame
df_nilai = pd.DataFrame(data_nilai)

# 3. Hitung nilai akhir (UTS 40% + UAS 60%)
df_nilai["nilai_akhir"] = (df_nilai["uts"] * 0.4) + (df_nilai["uas"] * 0.6)

print(df_nilai)

# ---------------------------------------------------------------------------------

# Jawaban Soal 2 di sini
# Filter mahasiswa yang nilai_akhir > 80
mahasiswa_lulus = df_nilai[df_nilai["nilai_akhir"] > 80]

print("\n================================")
print("\n",mahasiswa_lulus)
```
